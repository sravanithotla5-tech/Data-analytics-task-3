```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [12000, 15000, 18000, 14000, 20000, 22000],
    'Profit': [2000, 3000, 4000, 2500, 5000, 6000],
    'Customers': [100, 120, 150, 130, 180, 200]
}

df = pd.DataFrame(data)

df
```

```python
plt.figure(figsize=(8, 5))
plt.bar(df['Month'], df['Sales'])
plt.title('Monthly Sales Bar Chart')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Customers'], df['Sales'])
plt.title('Customers vs Sales')
plt.xlabel('Customers')
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Sales'], bins=5, edgecolor='black')
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df[['Sales', 'Profit', 'Customers']])
plt.title('Sales, Profit and Customers Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(df[['Sales', 'Profit', 'Customers']].corr(), annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Customers', y='Sales', size='Profit')
plt.title('Customers vs Sales')
plt.show()
```

```python
sns.pairplot(df[['Sales', 'Profit', 'Customers']])
plt.show()
```

```python
fig = px.line(df, x='Month', y='Sales', title='Interactive Monthly Sales')
fig.show()
```

```python
fig = px.bar(df, x='Month', y='Profit', title='Interactive Monthly Profit')
fig.show()
```

```python
fig = px.scatter(
    df,
    x='Customers',
    y='Sales',
    size='Profit',
    hover_name='Month',
    title='Interactive Customers vs Sales'
)
fig.show()
```

```python
fig = px.pie(
    df,
    names='Month',
    values='Sales',
    title='Monthly Sales Distribution'
)
fig.show()
```

```python
fig = px.histogram(
    df,
    x='Sales',
    nbins=5,
    title='Interactive Sales Distribution'
)
fig.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Line Chart
axes[0, 0].plot(df['Month'], df['Sales'], marker='o')
axes[0, 0].set_title('Monthly Sales')

# Bar Chart
axes[0, 1].bar(df['Month'], df['Profit'])
axes[0, 1].set_title('Monthly Profit')

# Scatter Plot
axes[1, 0].scatter(df['Customers'], df['Sales'])
axes[1, 0].set_title('Customers vs Sales')

# Histogram
axes[1, 1].hist(df['Sales'], bins=5, edgecolor='black')
axes[1, 1].set_title('Sales Distribution')

plt.tight_layout()
plt.show()
```

```python
df.to_csv('sales_data.csv', index=False)
print("CSV file created successfully!")
```
